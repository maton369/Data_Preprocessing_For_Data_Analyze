```python
# OpenCV-python のインポート
import cv2
import numpy as np
```

```python
path = "../data"

# OpenCV-pythonのインポート
# お試しで２枚読み込ませます
img_1 = cv2.imread(path + "/広島風お好み焼き.jpg", cv2.IMREAD_COLOR)
img_2 = cv2.imread(path + "/いちごとシャンパン.jpg", cv2.IMREAD_COLOR)
```

```python
print(type(img_1))
# 多次元配列であることがわかります

print(img_1.shape)
print(img_1)
```

```python
import matplotlib.pyplot as plt
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("広島風お好み焼き")
plt.show()
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_2, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("いちごとシャンパン")
plt.show()
```
