```python
# OpenCV-python のインポート
import cv2
import numpy as np
```

```python
path = "../data"

# OpenCV-pythonのインポート
# お試しで２枚読み込ませます
img_1 = cv2.imread(path + "/広島風お好み焼き.jpg", cv2.IMREAD_COLOR)
img_2 = cv2.imread(path + "/いちごとシャンパン.jpg", cv2.IMREAD_COLOR)
```

```python
print(type(img_1))
# 多次元配列であることがわかります

print(img_1.shape)
print(img_1)
```

```python
import matplotlib.pyplot as plt
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("広島風お好み焼き")
plt.show()
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_2, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("いちごとシャンパン")
plt.show()
```

```python
# グレースケール変換
img_1_gray = cv2.cvtColor(img_1, cv2.COLOR_BGR2GRAY)

# matplotlibで表示
plt.imshow(img_1_gray, cmap="gray")
plt.axis("off")
plt.title("広島風お好み焼き（グレースケール）")
plt.show()
```

```python
# BGRからRGBへ変換（方法①：チャンネル順を逆にするだけ）
img_1_RGB_1 = img_1[:, :, ::-1]

# BGRからRGBへ変換（方法②：OpenCV公式のカラースペース変換）
img_1_RGB_2 = cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB)

# 表示（2枚並べる）
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_1_RGB_1)
plt.axis("off")
plt.title("方法①: スライス")

plt.subplot(1, 2, 2)
plt.imshow(img_1_RGB_2)
plt.axis("off")
plt.title("方法②: cvtColor")

plt.show()
```

```python
# BGRのまま表示（色がずれるはず）
plt.imshow(img_1)
plt.axis("off")
plt.title("BGRのまま表示（色ズレ）")
plt.show()

# RGBに変換して表示（自然な色になる）
plt.imshow(cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("RGBに変換後（正しい色）")
plt.show()
```
