```python
# OpenCV-python のインポート
import cv2
import numpy as np
```

```python
path = "../data"

# OpenCV-pythonのインポート
# お試しで２枚読み込ませます
img_1 = cv2.imread(path + "/広島風お好み焼き.jpg", cv2.IMREAD_COLOR)
img_2 = cv2.imread(path + "/いちごとシャンパン.jpg", cv2.IMREAD_COLOR)
```

```python
print(type(img_1))
# 多次元配列であることがわかります

print(img_1.shape)
print(img_1)
```

```python
import matplotlib.pyplot as plt
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("広島風お好み焼き")
plt.show()
```

```python
# BGR → RGB に変換して表示
plt.imshow(cv2.cvtColor(img_2, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("いちごとシャンパン")
plt.show()
```

```python
# グレースケール変換
img_1_gray = cv2.cvtColor(img_1, cv2.COLOR_BGR2GRAY)

# matplotlibで表示
plt.imshow(img_1_gray, cmap="gray")
plt.axis("off")
plt.title("広島風お好み焼き（グレースケール）")
plt.show()
```

```python
# BGRからRGBへ変換（方法①：チャンネル順を逆にするだけ）
img_1_RGB_1 = img_1[:, :, ::-1]

# BGRからRGBへ変換（方法②：OpenCV公式のカラースペース変換）
img_1_RGB_2 = cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB)

# 表示（2枚並べる）
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_1_RGB_1)
plt.axis("off")
plt.title("方法①: スライス")

plt.subplot(1, 2, 2)
plt.imshow(img_1_RGB_2)
plt.axis("off")
plt.title("方法②: cvtColor")

plt.show()
```

```python
# BGRのまま表示（色がずれるはず）
plt.imshow(img_1)
plt.axis("off")
plt.title("BGRのまま表示（色ズレ）")
plt.show()

# RGBに変換して表示（自然な色になる）
plt.imshow(cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("RGBに変換後（正しい色）")
plt.show()
```

```python
# pyplotでは、では、cv2と色空間が逆（RGB空間）なので、cv2.cvtColorで色空間を変換してから表示させます。
src_1 = cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB)

plt.imshow(src_1)
# plt.imshow(img_1)
# 挙動が異なる
# cv2_imshow(src_1)
```

```python
# 画像の一部を切り抜き
trim_src_1 = src_1[800:2800, 2300:4300]
plt.imshow(trim_src_1)
```

```python
# 画像の表示関数を定義（OpenCVはBGRなのでRGBに変換して表示）
def show_image(img, title=""):
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis("off")
    plt.show()


# 元画像を表示
show_image(img_1, "Original")

# 上下反転
img_1_flip_ud = cv2.flip(img_1, 0)
show_image(img_1_flip_ud, "Flipped UD")

# 画像を保存
cv2.imwrite(path + "/save.jpeg", img_1_flip_ud)
```

```python
# 元画像
show_image(img_2, "Original")

# 左右反転
img_2_flip_lr = cv2.flip(img_2, 1)
show_image(img_2_flip_lr, "Flipped LR")
```

```python
# ブラー処理
img_1_blur_1 = cv2.blur(src=img_1, ksize=(1, 1))
img_1_blur_2 = cv2.blur(src=img_1, ksize=(5, 5))
img_1_blur_3 = cv2.blur(src=img_1, ksize=(10, 10))

# 1枚ずつ表示
show_image(img_1, "Original")
show_image(img_1_blur_1, "Blur (1x1)")
show_image(img_1_blur_2, "Blur (5x5)")
show_image(img_1_blur_3, "Blur (10x10)")
```
