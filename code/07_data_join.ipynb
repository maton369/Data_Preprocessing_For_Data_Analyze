```python
import pandas as pd
```

```python
transaction_history = pd.read_csv(
    "../data/transaction_history.csv", dtype={"customer_id": str}
)  # 顧客の購入履歴データ
product_master = pd.read_csv("../data/product_master.csv")  # 商品マスターのデータ
```

```python
# transaction_historyとProduct_masterをproduct_idで結合（Left Join)し、marged_dataオブジェクトに格納
merged_data = transaction_history.merge(product_master, on="product_id", how="left")
```
